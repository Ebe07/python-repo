```python
#trip information
trip_name = "roadtrip 2026"
driver_name = "evelyn"
car_model = "shelby Mustang"
start_city = "seattle"
end_city = "los Angeles"

#lists
destinations = ["seattle", "portland", "San Francisco", "Los Angelos"]
distances = [175, 635, 380]
packing_list = ["sunglasses", "camera", "snacks", "wattter bottle", "playlists"]

#gas
total_miles = sum(distances)

MPG = 35
price_per_gallon = 1.99
gas_cost = (total_miles / MPG) * price_per_gallon

print("-" * 30)
print(f"Trip: {trip_name.upper()}")
print(f"Driver:{driver_name}")
print(f"Vehical: {car_model}")

print(f"first stop: {destinations[0]}")

print(f"Last stop: {destinations[-1]}")

print(f"total distance: {total_miles} miles")
print(f"estimated gas cost: $ {gas_cost:.2f}")

item_count = len(packing_list)
print(f"items to pack: {item_count}")
print(f"Dont forget your {packing_list[2].lower()}")
print("-" * 30)
```
